### 2.5 Target Distribution

```python
train["Churn"] = train["Churn"].map({"Yes": 1, "No": 0})
```

```python
data_length = len(train)
churn = train["Churn"].sum()
not_churn = data_length - churn

churn_percen = round((churn / data_length) * 100, 2)
not_churn_percen = round((not_churn / data_length) * 100, 2)

print("Churn: ", churn)
print("Not-churn: ", not_churn)
print("Churn Percentage: ", churn_percen, "%")
print("Not-churn Percentage:", not_churn_percen, "%")
```
